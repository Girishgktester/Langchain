```python
# Load ENV file
import os

from dotenv import load_dotenv
load = load_dotenv(".env")
print("LANGSMITH_API_KEY is configured:", bool(os.getenv("LANGSMITH_API_KEY")))
print("LANGSMITH_TRACING is enabled:", bool(os.getenv("LANGSMITH_TRACING")))
```

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    base_url="http://localhost:11434",
    model="qwen3:8b",
    temperature=0.5,
    num_predict=2202,
    reasoning=False

)

response = llm.invoke("Hello India, how are you doing today?")
print(response.content)
```

```python

from langchain_core.prompts import ChatPromptTemplate, HumanMessagePromptTemplate, SystemMessagePromptTemplate

systemMessage = SystemMessagePromptTemplate.from_template("you are and 10+ years experianced SDET")
humansMessage = HumanMessagePromptTemplate.from_template("What is {env}")

prompt_template = ChatPromptTemplate([
   systemMessage,
   humansMessage
])

chain = prompt_template | llm 
response = chain.invoke({"env": "LLM explain me  in 10 words"}).content
response
```

```python
# String parser


from langchain_core.prompts import ChatPromptTemplate, HumanMessagePromptTemplate, SystemMessagePromptTemplate

from langchain_core.output_parsers import StrOutputParser

systemMessage = SystemMessagePromptTemplate.from_template("you are and 10+ years experianced SDET")
humansMessage = HumanMessagePromptTemplate.from_template("What is {env}")

prompt_template = ChatPromptTemplate([
   systemMessage,
   humansMessage
])

chain = prompt_template | llm | StrOutputParser()
response = chain.invoke({"env": "Appium exmlain me in detail"})
response
```
