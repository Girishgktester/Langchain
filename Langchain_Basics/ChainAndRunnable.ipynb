```python
# Chaining multiple chains

from langchain_core.prompts import (
    ChatPromptTemplate,
    HumanMessagePromptTemplate,
    SystemMessagePromptTemplate,
)

from langchain_core.output_parsers import StrOutputParser

systemMessage = SystemMessagePromptTemplate.from_template("you are and 10+ years experianced SDET")
humansMessage = HumanMessagePromptTemplate.from_template("What is {env}")

prompt_template = ChatPromptTemplate([systemMessage, humansMessage])

detailedResposeChain = prompt_template | llm | StrOutputParser()

headinInfoTempletae = ChatPromptTemplate.from_template("""Anayalise the respose and get me just heading from the {response} 
                                         response should in bullet points
                                         """)

chainWithHeading = {"response": detailedResposeChain } | headinInfoTempletae | llm | StrOutputParser()

response = chainWithHeading.invoke({"env": "explain me Appiumu in 20 lines"})
print(response)


```

```python
from langchain_core.prompts import (
    ChatPromptTemplate,
    HumanMessagePromptTemplate,
    SystemMessagePromptTemplate
)
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel

systemMessage = SystemMessagePromptTemplate.from_template( "You are a 10+ years experienced SDET")

humansMessage = HumanMessagePromptTemplate.from_template( "What is {env}")

prompt_template = ChatPromptTemplate([systemMessage, humansMessage])

# Chain 1 - Detailed explanation
detailedResponseChain = ( prompt_template| llm| StrOutputParser())

# Chain 2 - Short explanation
shortResponseTemplate = ChatPromptTemplate.from_template(
    """
    Explain {env} in 5 short bullet points.
    """)

shortResponseChain = (shortResponseTemplate | llm2 | StrOutputParser())

# Run both chains in parallel
parallelRunnable = RunnableParallel(detailed=detailedResponseChain, short=shortResponseChain)

response = parallelRunnable.invoke({ "env": "Appium"})

print("DETAILED RESPONSE:")
print(response["detailed"])

print("\nSHORT RESPONSE:")
print(response["short"])
```

```python
# Chaining multiple chains

from langchain_core.prompts import (
    ChatPromptTemplate,
    HumanMessagePromptTemplate,
    SystemMessagePromptTemplate,
)

from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel , RunnableLambda , chain

systemMessage = SystemMessagePromptTemplate.from_template("you are and 10+ years experianced SDET")
humansMessage = HumanMessagePromptTemplate.from_template("What is {env}")

prompt_template = ChatPromptTemplate([systemMessage, humansMessage])

detailedResposeChain = prompt_template | llm | StrOutputParser()

headinInfoTempletae = ChatPromptTemplate.from_template("""Anayalise the respose and get me just heading from the {response} 
                                         response should in bullet points
                                         """)
@chain
def choose_llm(response):
    response_text = str(response)

    if len(response_text) > 300:
        return llm2

    return llm

# llm_selector = RunnableLambda(choose_llm)

chainWithHeading = {"response": detailedResposeChain } | headinInfoTempletae | choose_llm | StrOutputParser()

response = chainWithHeading.invoke({"env": "explain me Appiumu in 20 lines"})
print(response)

```
