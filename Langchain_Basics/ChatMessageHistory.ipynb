```python
# Chaining multiple chains

from langchain_core.prompts import (ChatPromptTemplate,)

from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import chain
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_community.chat_message_histories import ChatMessageHistory


# prompt with history
template = ChatPromptTemplate.from_messages([
    ('human', "{prompt}"),
    ('placeholder', "{history}")

])


# chain that prompt and call llm and parse
chain = template | llm | StrOutputParser()

# get the session id if exists return it or else assign...

store = {}
def get_message_History(session_id:str) -> BaseChatMessageHistory:
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]

# for history using runnable,messagehistory passing chain that chain- llm then session id to remember who it is
# prompt and histruy 
history = RunnableWithMessageHistory(chain, get_message_History, input_messages_key="prompt", history_messages_key="history")
# session id is GK
session_id = "Gk"

# invoke the prompt from histry  and need to configure the session id to remeber the histry with session id help
response1 = history.invoke({"prompt": "distance between earth and sun?"}, config={"configurable": {"session_id": session_id}})
response2 = history.invoke({"prompt": "how about moon?"}, config={"configurable": {"session_id": session_id}})

print(response1)
print(response2)
```

```python
# Persistent chat history with SQLite

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_community.chat_message_histories import SQLChatMessageHistory


def get_session_history(session_id):
    return SQLChatMessageHistory(
        session_id=session_id,
        connection="sqlite:///chathistory.db",
    )


# Prompt with history
template = ChatPromptTemplate.from_messages([
    ("human", "{prompt}"),
    ("placeholder", "{history}"),
])

chain = template | llm | StrOutputParser()

history = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="prompt",
    history_messages_key="history",
)

session_id = "Gk"
get_session_history(session_id).clear()

response1 = history.invoke(
    {"prompt": "distance between earth and sun?"},
    config={"configurable": {"session_id": session_id}},
)
response2 = history.invoke(
    {"prompt": "how about moon?"},
    config={"configurable": {"session_id": session_id}},
)

print(response1)
print(response2)
```
